## Study question and data

**Study question:** How does cogeneration potential vary with the site's operating periods?

## Step 1: Run the period reference and default model

This cell loads `zonal_site_multiperiod.json` and runs `all_periods.all_heat_integration()` for every period. `all_periods.cogeneration()` then applies the default turbine model per period, giving `period_cogeneration`. `weighted` is the summary with the weighted average across periods.

```python
from OpenPinch import PinchProblem

problem = PinchProblem("zonal_site_multiperiod.json", project_name="Site")
problem.target.all_periods.all_heat_integration()
period_cogeneration = problem.target.all_periods.cogeneration()
weighted = problem.summary_frame(include_weighted_average=True)
weighted
```

## Step 2: Compare period turbine models

This cell runs the Sun–Smith, small and large Varbanov, and isentropic turbine models for every period through `target.all_periods`. Each Varbanov model warns if a stage's work falls outside its 2 MW band. The isentropic model uses `efficiency=0.8`. Compare `sun_smith`, `varbanov_small`, `varbanov_large` and `isentropic` period by period. The last line lists the identifiers in `problem.period_results`.

```python
sun_smith = problem.target.all_periods.sun_smith_cogeneration()
varbanov_small = problem.target.all_periods.varbanov_small_turbine_cogeneration()
varbanov_large = problem.target.all_periods.varbanov_large_turbine_cogeneration()
isentropic = problem.target.all_periods.isentropic_cogeneration(efficiency=0.8)
list(problem.period_results)
```

## Review the result

Review the weighted table with all period method results to identify which operating conditions drive annual cogeneration value.

```python
from IPython.display import display

period_cogeneration_methods = {
    "default": period_cogeneration,
    "Sun-Smith": sun_smith,
    "Varbanov small": varbanov_small,
    "Varbanov large": varbanov_large,
    "isentropic": isentropic,
}
display(weighted)
display(period_cogeneration_methods)
```

## Interpret the result

Check the period with the lowest viable steam flow as well as the weighted annual opportunity.
