## Study question and data

**Study question:** How sensitive is cogeneration potential to the turbine model used for screening?

## Step 1: Establish the cogeneration reference

This cell loads `pulp_mill.json`, runs `all_heat_integration()` and then `total_site_heat_integration()` as `base`. `cogeneration(base_target=base)` runs the default turbine model on that Total Site target. The result is `default`, and `cogeneration_summary` tabulates the solved targets.

```python
from OpenPinch import PinchProblem

problem = PinchProblem("pulp_mill.json", project_name="Site")
problem.target.all_heat_integration()
base = problem.target.total_site_heat_integration()
default = problem.target.cogeneration(base_target=base)
cogeneration_summary = problem.summary_frame()
```

## Step 2: Compare named turbine models

This cell runs the three named turbine models. `sun_smith_cogeneration()` and `varbanov_cogeneration()` select their models by method name. `isentropic_cogeneration(efficiency=0.8)` sets the turbine efficiency explicitly. Each call takes the previous result as `base_target`. Compare `sun_smith`, `varbanov` and `isentropic` with `default`.

```python
sun_smith = problem.target.sun_smith_cogeneration(base_target=default)
varbanov = problem.target.varbanov_cogeneration(base_target=sun_smith)
isentropic = problem.target.isentropic_cogeneration(
    efficiency=0.8, base_target=varbanov
)
```

## Review the result

Compare the common process summary with the alternative cogeneration results to see how utility conditions and turbine assumptions change recoverable power.

```python
from IPython.display import display

cogeneration_methods = {
    "default": default,
    "Sun-Smith": sun_smith,
    "Varbanov": varbanov,
    "isentropic": isentropic,
}
display(cogeneration_summary)
display(cogeneration_methods)
```

## Interpret the result

Compare models against the same Total Site base target and state efficiency assumptions beside the reported power potential.
